```python
print("name: B Harshitha Pai  USN1WN24CS059")
def calculate_cost(state):

    attacks = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):

            if abs(state[i] - state[j]) == abs(i - j):
                attacks += 1
    return attacks

def hill_climbing_4_queens(initial_state):
    current_state = list(initial_state)
    current_cost = calculate_cost(current_state)

    print(f"Initial State: {current_state} | Cost: {current_cost}\n")

    step = 1
    while True:
        best_neighbor = None
        best_cost = current_cost
        chosen_pair = None

        print(f"--- Step {step}: Evaluating Neighbors ---")


        for i in range(len(current_state)):
            for j in range(i + 1, len(current_state)):

                neighbor = list(current_state)
                neighbor[i], neighbor[j] = neighbor[j], neighbor[i]
                neighbor_cost = calculate_cost(neighbor)

                print(f"Swap ({i},{j}) -> Neighbor: {neighbor} | Cost: {neighbor_cost}")


                if neighbor_cost < best_cost:
                    best_cost = neighbor_cost
                    best_neighbor = neighbor
                    chosen_pair = (i, j)


        if best_neighbor is None:
            print(f"\nTerminated: Local minimum reached at state {current_state} with cost {current_cost}")
            break


        current_state = best_neighbor
        current_cost = best_cost
        print(f"\n[Moved] Swapped {chosen_pair} -> New Current State: {current_state} | Cost: {current_cost}\n")

        if current_cost == 0:
            print(f"Goal Configuration Reached successfully: {current_state} ")
            break

        step += 1


initial_state = [3, 1, 2, 0]
hill_climbing_4_queens(initial_state)
```

```python
import random
import math

print("name: B Harshitha Pai  USN:1WN24CS059")
def get_conflicts(board):

    conflicts = 0
    n = len(board)
    for i in range(n):
        for j in range(i + 1, n):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def get_neighbor(board):

    neighbor = list(board)
    col = random.randint(0, 7)
    row = random.randint(0, 7)
    while neighbor[col] == row:
        row = random.randint(0, 7)
    neighbor[col] = row
    return neighbor

def simulated_annealing():

    current_board = [random.randint(0, 7) for _ in range(8)]
    current_energy = get_conflicts(current_board)

    temperature = 1.0
    cooling_rate = 0.995
    min_temperature = 0.00001

    print(f"Starting Board: {current_board} with {current_energy} conflicts.")

    while temperature > min_temperature and current_energy > 0:
        neighbor = get_neighbor(current_board)
        neighbor_energy = get_conflicts(neighbor)
        d_energy = neighbor_energy - current_energy

        if d_energy < 0:
            current_board = neighbor
            current_energy = neighbor_energy
        else:
            acceptance_probability = math.exp(-d_energy / temperature)
            if random.random() < acceptance_probability:
                current_board = neighbor
                current_energy = neighbor_energy

        temperature *= cooling_rate

    return current_board, current_energy

solution, conflicts = simulated_annealing()

print("\n--- Final Result ---")
print(f"Solution Board: {solution}")
print(f"Remaining Conflicts: {conflicts}")
if conflicts == 0:
    print("Success! A valid solution was found.")
else:
    print("Failed to find a perfect solution before cooling down. Try running again!")
```
