```python
facts = {
    ("American", "Robert"),
    ("Enemy", "A", "America"),
    ("Owns", "A", "T1"),
    ("Missile", "T1")
}
print("B HARSHITHA PAI")
print("1WN24CS059")
print("Initial Facts:")
for f in sorted(facts):
    print(f[0] + "(" + ", ".join(f[1:]) + ")")

iteration = 1
new = True

while new:
    new = False
    new_facts = set()

    for f in facts:
        if f[0] == "Missile":
            new_facts.add(("Weapon", f[1]))

    if ("Enemy", "A", "America") in facts:
        new_facts.add(("Hostile", "A"))

    for f in facts:
        if f[0] == "Missile":
            x = f[1]
            if ("Owns", "A", x) in facts:
                new_facts.add(("Sells", "Robert", x, "A"))

    for f in facts:
        if f[0] == "American":
            p = f[1]

            for w in facts:
                if w[0] == "Weapon":
                    q = w[1]

                    for s in facts:
                        if s[0] == "Sells" and s[1] == p and s[2] == q:
                            r = s[3]

                            if ("Hostile", r) in facts:
                                new_facts.add(("Criminal", p))

    new_facts = new_facts - facts

    if new_facts:
        print("\nIteration", iteration, ":")
        for f in sorted(new_facts):
            print(f[0] + "(" + ", ".join(f[1:]) + ")")

        facts.update(new_facts)
        new = True
        iteration += 1

print("\nFinal Facts:")
for f in sorted(facts):
    print(f[0] + "(" + ", ".join(f[1:]) + ")")

if ("Criminal", "Robert") in facts:
    print("\nConclusion: Robert is a Criminal.")
```
