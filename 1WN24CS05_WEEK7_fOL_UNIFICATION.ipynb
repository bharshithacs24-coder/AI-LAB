```python
print("name:B Harshitha Pai   USN:1WN24CS059")
def unify(x, y, subst):

    if x == y:
        return subst

    if x.islower():
        subst[x] = y
        return subst

    elif y.islower():
        subst[y] = x
        return subst

    else:
        return None




print("Unification of Eats(x, Apple) and Eats(Riya, y)")

subst = {}

result1 = unify("x", "Riya", subst)
result2 = unify("y", "Apple", subst)

if result1 is not None and result2 is not None:
    print("Unification Successful")
    print("Substitution:", subst)
    print("Unified Expression: Eats(Riya, Apple)")
else:
    print("Unification Failed")




```
